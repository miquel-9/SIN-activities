**Mètode que implementa una nova heurística**

En aquesta cas la nova heurística és simplement multiplicar per 2 la distància de Manhattan

```python
# Nova heurística a provar: 2 * distància de Manhattan
def nova_heuristica(state, end_state):
    """
    Heurística que multiplica per 2 la distància Manhattan 
    """
    return 2 * getManhattanDistance(state, end_state) 

# Una sola línia és suficient per a que run_algorithm i run_all_algorithms la coneguen.
register_heuristic("Nova Heurística", nova_heuristica)
```

**Exemple amb grandària 3**

Es defineix l'estat inicial i el final, s'executen tots els algorismes disponibles i es visualitzen els resultats

```python
# Grandària 3

size = 3  # Example for a 3x3 puzzle

initial_state_matrix = [
    [7, 8, 1],
    [4, 0, 6],
    [2, 3, 5]
]


end_state_matrix = [
    [1, 2, 3],
    [8, 0, 4],
    [7, 6, 5]
]

# Convertir la matriu en una llista unidimensional
initial_state = [num for row in initial_state_matrix for num in row]
# Convertir la matriu en una llista unidimensional
end_state = [num for row in end_state_matrix for num in row]

# Executar tots els algorismes i mostrar els resultats en una taula
results = {}
if isSolvable(initial_state, end_state):
    # Cridar a la funció per a executar tots els algorismes
    results = run_all_algorithms(initial_state, end_state, size, 40)
    show_results(results)
    plot_algorithm_comparison(results)
else:
    print("The puzzle is not solvable.")


```

**Exemple d'estudi d'admissibilitat i consistència**

Per a l'estat inicial anterior es compara el valor de la heurística amb el cost òptim obtingut mitjançant BFS. Després s'empra `check_consistency` per a cercar un possible contraexemple de consistència. Aquestes comprovacions sobre un únic exemple aporten evidència, però no constitueixen una demostració general.

```python
# Estudi d'admissibilitat per a l'estat inicial anterior
h_initial = nova_heuristica(initial_state, end_state)
h_goal = nova_heuristica(end_state, end_state)
optimal_cost = results['BFS']['cost']

print("Valor en l'estat inicial:", h_initial)
print("Valor en l'estat objectiu:", h_goal)
print("Cost òptim obtingut amb BFS:", optimal_cost)
print("És no negativa en aquest exemple?", h_initial >= 0)
print("Val cero en l'objectiu?", h_goal == 0)
print("No supera el cost òptim?", h_initial <= optimal_cost)

# Estudi de consistència des del mateix estat inicial
consistency_result = check_consistency(
    nova_heuristica, initial_state, end_state, size, max_states=10000
)
print("Resultat de la prova de consistència:")
print(consistency_result)
```
