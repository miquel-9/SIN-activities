# Anàlisi d'Algorismes de Cerca
Comparació estadística entre A* MD + LC i Voraz Manhattan sobre n puzles aleatoris.

Mètriques a comparar:
- Cost de la solució (nombre de moviments)
- Nodes expandits
- Temps d'execució
- Diferència entre solucions

```python
# Importar paquets necessaris
import search
from search import *
import time
import statistics
import matplotlib.pyplot as plt
```

## Configuració i Execució de l'Experiment

```python
# Configuració
size = 3  # Grandària del puzle
num_iterations = 1000  # Nombre de puzles aleatoris
end_state = generate_spiral_state(size)

# Estructures per a emmagatzemar resultats
astar_costs = []
astar_expanded = []
astar_times = []
greedy_costs = []
greedy_expanded = []
greedy_times = []
cost_differences = [] 
time_differences = [] 

print(f"Iniciant experiment amb {num_iterations} puzles {size}x{size}")
print("Estat objectiu:")
visualize_state(end_state)

# Bucle principal de l'experiment
for i in range(num_iterations):
    # Generar estat inicial aleatori
    initial_state = generate_random_state(size, end_state)
    
    if not isSolvable(initial_state, end_state):
        print(f"\nPuzle {i+1}: No té solució, generant un altre...")
        continue
        
    print(f"\rProcessant puzle {i+1}/{num_iterations}", end='', flush=True)
    
    # A* MD + LC

    result = run_algorithm('A (MD + LC)', initial_state, end_state, size)

    astar_result = result['A (MD + LC)']
    
    # Voraz Manhattan

    result = run_algorithm('Voraz (Manhattan)', initial_state, end_state, size)

    greedy_result = result['Voraz (Manhattan)']
    
    # Guardar resultats

    astar_costs.append(astar_result['cost'])
    astar_expanded.append(astar_result['nodes_expanded'])
    astar_times.append(astar_result['time'])
    
    greedy_costs.append(greedy_result['cost'])
    greedy_expanded.append(greedy_result['nodes_expanded'])
    greedy_times.append(greedy_result['time'])
    
    cost_diff = greedy_result['cost'] - astar_result['cost']
    cost_differences.append(cost_diff)
        
    time_diff = astar_result['time'] - greedy_result['time']
    time_differences.append(time_diff)
        


print("\nExperiment completat!")
```

## Anàlisi de Resultats

```python
# Calcular estadístiques
stats = {
    'A*': {
        'coste_min': min(astar_costs),
        'coste_max': max(astar_costs),
        'coste_medio': statistics.mean(astar_costs),
        'expandidos_medio': statistics.mean(astar_expanded),
        'tiempo_medio': statistics.mean(astar_times)
    },
    'Voraz': {
        'coste_min': min(greedy_costs),
        'coste_max': max(greedy_costs),
        'coste_medio': statistics.mean(greedy_costs),
        'expandidos_medio': statistics.mean(greedy_expanded),
        'tiempo_medio': statistics.mean(greedy_times)
    }
}

print("Estadístiques d'A* MD LC:")
print(f"Cost - Min: {stats['A*']['coste_min']}, Max: {stats['A*']['coste_max']}, Mitjana: {stats['A*']['coste_medio']:.2f}")
print(f"Nodes expandits (mitjana): {stats['A*']['expandidos_medio']:.2f}")
print(f"Temps mitjà: {stats['A*']['tiempo_medio']:.3f}s\n")

print("Estadístiques de Voraç Manhattan:")
print(f"Cost - Min: {stats['Voraz']['coste_min']}, Max: {stats['Voraz']['coste_max']}, Mitjana: {stats['Voraz']['coste_medio']:.2f}")
print(f"Nodes expandits (mitjana): {stats['Voraz']['expandidos_medio']:.2f}")
print(f"Temps mitjà: {stats['Voraz']['tiempo_medio']:.3f}s\n")

print("Diferències de cost (Voraç - A*):")
print(f"Mínima: {min(cost_differences)}")
print(f"Màxima: {max(cost_differences)}")
print(f"Mitjana: {statistics.mean(cost_differences):.2f}")
```

## Visualització de Resultats

```python
# Gràfic de caixes per a costos
plt.figure(figsize=(10, 6))
plt.boxplot([astar_costs, greedy_costs])
plt.title('Distribució de Costos per Algorisme')
plt.ylabel('Nombre de moviments')
plt.xticks([1, 2], ['A*', 'Voraç'])  # ← Aquí especifiques les etiquetes
plt.show()


# Gràfic de barres per a mitjanes
metrics = ['Cost Mitjà', 'Nodes Expandits\n(mitjana)', 'Temps Mitjà (s)']
astar_values = [stats['A*']['coste_medio'], stats['A*']['expandidos_medio'], stats['A*']['tiempo_medio']]
greedy_values = [stats['Voraz']['coste_medio'], stats['Voraz']['expandidos_medio'], stats['Voraz']['tiempo_medio']]

plt.figure(figsize=(12, 6))
x = range(len(metrics))
width = 0.35

plt.bar([i - width/2 for i in x], astar_values, width, label='A*', color='skyblue')
plt.bar([i + width/2 for i in x], greedy_values, width, label='Voraz', color='lightcoral')

plt.xlabel('Mètriques')
plt.ylabel('Valors')
plt.title('Comparació de Mètriques entre Algorismes')
plt.xticks(x, metrics)
plt.legend()
plt.show()

# Histograma de diferències de cost
plt.figure(figsize=(10, 6))
plt.hist(cost_differences, bins=20, edgecolor='black')
plt.title('Distribució de Diferències de Cost (Voraç - A*)')
plt.xlabel('Diferència en nombre de moviments')
plt.ylabel('Freqüència')
plt.show()

# Histograma de diferències de temps
plt.figure(figsize=(10, 6))
plt.hist(time_differences, bins=100, edgecolor='black')
plt.title('Distribució de Diferències de Temps (A* - Voraç)')
plt.xlabel('Diferència en unitats de temps')
plt.ylabel('Freqüència')
plt.show()
```
